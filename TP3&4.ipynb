## Exercice 1 : Arbres d'expression

### 4.

```python
addition = noeud.Noeud("+")
addition.ajouter_noeud(noeud.Noeud(2))
addition.ajouter_noeud(noeud.Noeud("y"))

expression = noeud.Noeud("exp")
expression.ajouter_noeud(addition)

print(expression.afficher_expression())
```

### 5.

```python
print(expression.evaluer({"y": 1}))

try:
    expression.evaluer({})
except ValueError as erreur:
    print(erreur)
```

### 6.

```python
expression.tracer("y", [i / 10 for i in range(-50, 21)])
```

### 7.

```python
a_simplifier = noeud.Noeud("+", [noeud.Noeud("*", [noeud.Noeud(1), noeud.Noeud("x")]), noeud.Noeud("+", [noeud.Noeud(1), noeud.Noeud("exp", [noeud.Noeud(0)])])])

print(a_simplifier.afficher_expression())
print(a_simplifier.simplifiee().afficher_expression())
```

### 8.

```python
a_deriver = noeud.Noeud("*", [noeud.Noeud(3), noeud.Noeud("sin", [noeud.Noeud("x")])])

print(a_deriver.derivee("x").afficher_expression())
print(a_deriver.derivee("x").simplifiee().afficher_expression())
```
